### **Task Manager — Python Project**

This project is a command-line task manager built step by step in Python. The goal is to create a fully interactive application that allows the user to add, view, edit, delete and organize personal tasks — each defined by a title, a priority level, a status and a deadline. Tasks are persisted between sessions using a JSON file, can be sorted by priority, tracked through a weighted productivity score, and exported to an Excel spreadsheet using pandas.

**Adding a Single Task**

This cell initializes an empty task list, prompts the user to fill in four fields — title, priority, status, and deadline — stores them in a dictionary, and appends it to the list before printing the result.

```python
tasks = []

task = {
  "title": input("Title: "),
  "priority": input("Priority (low/medium/high): "),
  "status": input("Status (todo/in progress/done): "),
  "deadline": input("Deadline (YYYY-MM-DD): ")
}

tasks.append(task)
print(tasks)
```

**Adding Multiple Tasks**

This cell runs a continuous loop that keeps prompting the user to enter tasks one by one — each with a title, priority, status, and deadline. Typing "stop" as the title breaks the loop and prints the final task list.

```python
while True:
    title = input("Title (or type stop): ")

    if title == "stop":
        break

    priority = input("Priority (low/medium/high): ")
    status = input("Status (todo/in progress/done): ")
    deadline = input("Deadline (YYYY-MM-DD): ")

    task = {
        "title": title,
        "priority": priority,
        "status": status,
        "deadline": deadline
    }

    tasks.append(task)

print(tasks)
```

**Marking a Task as Done**

This cell displays all tasks with their index and current status, asks the user to pick one by number, and updates its status to "done" — directly modifying the dictionary inside the list.

```python

for i, task in enumerate(tasks):
    print(i, task["title"], "-", task["status"])
index = int(input("Which task is done? "))

tasks[index]["status"] = "done"

print(tasks)
```

**Deleting a Task**

This cell displays all tasks with their index and status, asks the user to pick one by number, and permanently removes it from the list using pop() — automatically shifting the indexes of all remaining tasks.

```python
for i, task in enumerate(tasks):
    print(i, task["title"], "-", task["status"])

index = int(input("Task to delete: "))

tasks.pop(index)

print(tasks)
```

**Loading Tasks from JSON**

This cell opens the tasks.json file and loads its content into the task list — restoring all previously saved tasks so they can be used in the current session.

```python
import json
```

```python
with open("tasks.json", "r") as file:
    tasks = json.load(file)
print (tasks)
```

**Saving Tasks to JSON**

This cell opens tasks.json in write mode and uses json.dump() to serialize the tasks list — converting the Python list of dictionaries into JSON format and writing it to the file, permanently saving all current tasks for future sessions.

```python

with open("tasks.json", "w") as file:
    json.dump(tasks, file)
```

**Interactive Task Manager Menu**

This cell loads the tasks from the **tasks.json** file at startup, then runs a continuous loop that displays an **8-option** menu allowing to add, view, edit, delete, and sort tasks, track  productivity, and save everything back to JSON before exiting.

```python
# Charger les tâches au démarrage
with open("tasks.json", "r") as file:
    tasks = json.load(file)

while True:

    print("\nTASK MANAGER")
    print("1 Add task")
    print("2 Show tasks")
    print("3 Mark as done")
    print("4 Delete task")
    print("5 Edit task")
    print("6 Show by priority ")
    print("7 My productivity")
    print("8 Exit")

    choice = input("Choice: ")

    if choice == "1":
        title = input("Title: ")
        priority = input("Priority (low/medium/high): ")
        status = "todo"
        deadline = input("Deadline (YYYY-MM-DD): ")

        task = {
            "title": title,
            "priority": priority,
            "status": status,
            "deadline": deadline
        }
        tasks.append(task)

    elif choice == "2":
        for i, task in enumerate(tasks):
            print(i, task["title"], "-", task["priority"], "-", task["status"], "-", task["deadline"])

    elif choice == "3":
        for i, task in enumerate(tasks):
            print(i, task["title"], "-", task["status"])
        index = int(input("Task number: "))
        tasks[index]["status"] = "done"

    elif choice == "4":
        for i, task in enumerate(tasks):
            print(i, task["title"], "-", task["status"])
        index = int(input("Task to delete: "))
        tasks.pop(index)

    elif choice == "5":
        for i, task in enumerate(tasks):
            print(i, task["title"], "-", task["priority"], "-", task["status"], "-", task["deadline"])
        index = int(input("Task to edit: "))
        print("Leave blank to keep current value")

        title = input(f"Title ({tasks[index]['title']}): ")
        priority = input(f"Priority ({tasks[index]['priority']}): ")
        status = input(f"Status ({tasks[index]['status']}): ")
        deadline = input(f"Deadline ({tasks[index]['deadline']}): ")

        if title: tasks[index]["title"] = title
        if priority: tasks[index]["priority"] = priority
        if status: tasks[index]["status"] = status
        if deadline: tasks[index]["deadline"] = deadline

    elif choice == "6":
      priority_order = {"high": 1, "medium": 2, "low": 3}

    # Séparer les tâches actives et terminées
      active_tasks = [task for task in tasks if task["status"] != "done"]
      done_tasks   = [task for task in tasks if task["status"] == "done"]

    # Trier chaque groupe par priorité puis par deadline
      active_sorted = sorted(active_tasks, key=lambda task: (priority_order.get(task["priority"], 4), task["deadline"]))
      done_sorted   = sorted(done_tasks,   key=lambda task: (priority_order.get(task["priority"], 4), task["deadline"]))

    # Afficher les tâches actives
      print(f"\n{'#':<5}{'Title':<20}{'Priority':<12}{'Status':<15}{'Deadline'}")
      print("-" * 60)
      for i, task in enumerate(active_sorted):
        print(f"{i:<5}{task['title']:<20}{task['priority']:<12}{task['status']:<15}{task['deadline']}")

    # Afficher les tâches terminées en dessous
      if done_sorted:
        print(f"\n--- DONE ---")
        for i, task in enumerate(done_sorted):
            print(f"{i:<5}{task['title']:<20}{task['priority']:<12}{task['status']:<15}{task['deadline']}")

    elif choice == "7":
        total = len(tasks)

        if total == 0:
            print("No tasks found.")
        else:
            done = len([task for task in tasks if task["status"] == "done"])
            todo = total - done
            percentage = (done / total) * 100

            score = 0
            for task in tasks:
                if task["status"] == "done":
                    if task["priority"] == "high": score += 3
                    elif task["priority"] == "medium": score += 2
                    elif task["priority"] == "low": score += 1

            max_score = sum(3 if task["priority"] == "high" else 2 if task["priority"] == "medium" else 1 for task in tasks)
            productivity = (score / max_score) * 100 if max_score > 0 else 0

            print(f"\n PRODUCTIVITY REPORT")
            print("-" * 30)
            print(f"Total tasks    : {total}")
            print(f"Completed      : {done}")
            print(f"Remaining      : {todo}")
            print(f"Completion     : {percentage:.1f}%")
            print(f"Productivity   : {productivity:.1f}%")

    elif choice == "8":
        with open("tasks.json", "w") as file:
            json.dump(tasks, file)
        break
```

**Exporting Tasks to Excel**

This cell loads the saved tasks from the tasks.json file, converts them into a pandas DataFrame, and exports them to a tasks.xlsx file — making it easy to view and share the task list in a spreadsheet format compatible with Excel or Google Sheets.

```python
import pandas as pd

with open("tasks.json", "r") as file:
    tasks = json.load(file)

df = pd.DataFrame(tasks)
df.to_excel("tasks.xlsx", index=False)

print("Exported to tasks.xlsx")
```
